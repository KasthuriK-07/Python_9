```python
x=int(input("Enter x: "))
y=int(input("Enter y: "))
z=int(input("Enter z: "))
if x>y and x>z:
    print("X is greater")
elif y>x and y>z:
    print("Y is greater")
else:
    print("Z is greater")
```

```python
price=int(input("Enter price: "))
if price>1500 and price<=2000:#10% discount
    dis=price*0.10
    total_bill=price-dis
    print("Original Price: ",price)
    print("Discount Price: ",dis)
    print("How much Amount has to pay: ",total_bill)
else:
    pass#Empty stament
```

```python
x=int(input("Enter x: "))
print("1.Square")
print("2.Cube")
print("3.Exit")
ch=int(input("Enter the choice:"))
if ch==1:
    z=x**2
    print("Square value of",x,"is :",z)
elif ch==2:
    print("Cube value of",x,"is: ",x**3)
else:
    print("Quit")
```

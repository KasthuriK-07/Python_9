```python

def display_movies():
    print("\n----- AVAILABLE MOVIES -----")
    print("1. Leo")
    print("2. Jailer")
    print("3. GOAT")


def book_ticket():
    display_movies()

    movie_choice = int(input("\nSelect movie (1-3): "))

    if movie_choice == 1:
        movie = "Leo"
    elif movie_choice == 2:
        movie = "Jailer"
    elif movie_choice == 3:
        movie = "GOAT"
    else:
        print("Invalid movie selection!")
        return

    print("\nMovie selected:", movie)

    tickets = int(input("Enter number of tickets: "))

    if tickets <= 0:
        print("Please enter a valid ticket quantity!")
        return
    elif tickets > 10:
        print("Maximum 10 tickets allowed!")
        return

    price = 150
    total = tickets * price

    print("\n----- BOOKING SUMMARY -----")
    print("Movie:", movie)
    print("Tickets:", tickets)
    print("Price per ticket: Rs.", price)
    print("Total amount: Rs.", total)

    confirm = input("Confirm booking? (yes/no): ").lower()

    if confirm == "yes":
        print("\nTicket booking successful!")
        print("Enjoy your movie!")
        for i in range(1, tickets + 1):
            print("Ticket", i, "confirmed for", movie)
    else:
        print("Booking cancelled.")


def main():
    while True:
        print("\n===== MOVIE TICKET RESERVATION =====")
        print("1. View Movies")
        print("2. Book Tickets")
        print("3. Exit")

        choice = input("Enter your choice: ")

        if choice == "1":
            display_movies()
        elif choice == "2":
            book_ticket()
        elif choice == "3":
            print("Thank you for visiting!")
            break
        else:
            print("Invalid choice! Please try again.")


main()
```

```python
#OOP-Object Oriented Programming
#Class-User-Defined Prototype or collections of object or blueprint of an object
#class classname:-2 properties-attribute and behavior
class Student:
    university="VTU"#class variable
    def __init__(self,college):#Initialize constructor
        self.college=college#Instance variable
        print("Student Info ",college)
    #Methods-Inside the function to the body of the class
    def student_info(self,stud_name,stud_dept,stud_id):
        #self is the parameter to access method variable
        self.stud_name=stud_name
        self.stud_dept=stud_dept
        self.stud_id=stud_id
    def stud_marks(x,m1,m2,m3):
        x.m1=m1
        x.m2=m2
        x.m3=m3
        print("Student Marks: ",m1+m2+m3)
```

```python
#Inheritance
#Base or parent and child or derived class
#Derived class inherit base class properties
#Base class can't inherit derived class properties
#Single Inheritance-One base class and one derived class
class Student:#Base class
    university="VTU"#class variable
    def __init__(self,college):#Initialize constructor
        self.college=college#Instance variable
        print("Student Info ",college)
    #Methods-Inside the function to the body of the class
    def student_info(self,stud_name,stud_dept,stud_id):
        #self is the parameter to access method variable
        self.stud_name=stud_name
        self.stud_dept=stud_dept
        self.stud_id=stud_id
class Marks(Student):#Derived class
    def stud_marks(x,m1,m2,m3):
        x.m1=m1
        x.m2=m2
        x.m3=m3
        print("Student Marks: ",m1+m2+m3)
```

```python
s=Marks("Oxford")
s.student_info("Anu","IT",10)
s.stud_marks(90,67,45)
```

```python
#Multi-level inheritance
class Student:#Base class
    university="VTU"#class variable
    def __init__(self,college):#Initialize constructor
        self.college=college#Instance variable
        print("Student Info ",college)
    #Methods-Inside the function to the body of the class
    def student_info(self,stud_name,stud_dept,stud_id):
        #self is the parameter to access method variable
        self.stud_name=stud_name
        self.stud_dept=stud_dept
        self.stud_id=stud_id
class Marks(Student):#1st Derived class
    def stud_marks(x,m1,m2,m3):
        x.m1=m1
        x.m2=m2
        x.m3=m3
        print("Student Marks: ",m1+m2+m3)
class Performance(Marks):#2nd derived class
    def rank(self,grade):
        self.grade=grade
class Result(Performance):#3rd derived class
    def result(self,result):
        self.result=result
s=Result("Oxford")
s.student_info("Anu","IT",10)
s.stud_marks(90,67,45)
s.rank("A")
s.result("Pass")
print("Name: ",s.stud_name)
print("Total Marks: ",s.m1+s.m2+s.m3)
print("Rank: ",s.grade)
print("Result: ",s.result)
```

```python
#Multiple Inheritance-Multiple base class and only one derived class
class Add:#Base class
    def add(x,a,b):
        x.a=a
        x.b=b
        return a+b
class Sub:#Base class
    def sub(x,a,b):
        x.a=a
        x.b=b
        return a-b
class Div(Add,Sub):#Derived class
    def div(x,a,b):
        x.a=a
        x.b=b
        return a/b
```

```python
d=Div()
print(d.add(12,3))
print(d.sub(12,3))
print(d.div(12,3))
```

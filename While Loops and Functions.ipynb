```python
#Reverse the number
#789 =987
n=int(input("Enter number to reverse: "))
sum=0
while n!=0:#789!=0,78!=0,7!=0
    rem=n%10#789%10,rem=9,78%10=8
    sum=sum*10+rem#0*10+9,sum=9*10+8,sum=98*10+7=987
    n=n//10#Quotient-789/10=78,78//10=7
print(sum)
```

```python
#Armstrong Number
n=int(input("Enter number to reverse: "))
temp=n
len_n=len(str(n))
sum=0
while n!=0:#153!=0,15!=0
    rem=n%10#153%10,rem=3,rem=5
    sum=sum+rem**len_n#0+3**3,sum=27+5**3,sum=152+1**3,sum=153
    n=n//10#Quotient-153//10,n=15,n=1
print(n)
if temp==sum:#153==153
    print("Armstrong Number: ",sum)
else:
    print("Not Armstrong Number: ",sum)
```
